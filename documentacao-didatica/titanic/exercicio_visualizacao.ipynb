```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Carregando e limpando os dados
df = pd.read_csv('titanic.csv')
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['fare'] = pd.to_numeric(df['fare'], errors='coerce')

# 2. Idade por Sobrevivência
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', hue='survived', kde=True, element='step')
plt.title('Distribuição de Idade por Status de Sobrevivência')
plt.show()

# 3. Contagem por Gênero
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='sex', palette='pastel')
plt.title('Quantidade de Passageiros por Gênero')
plt.show()

# 4. Tarifa por Porto de Embarque
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='embarked', y='fare')
plt.ylim(0, 200) # Foco na maioria dos dados
plt.title('Preço da Passagem por Porto de Embarque')
plt.show()
```
