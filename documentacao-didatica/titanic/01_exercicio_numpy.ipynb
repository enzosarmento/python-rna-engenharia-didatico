# Exercício 01 — Análise Estatística Vetorial com `NumPy` (Titanic)

Agora que você conhece a velocidade vetorial do `NumPy` (`ndarray` — arranjo numérico homogêneo), vamos aplicá-la sobre os dados reais de idade (`age`) dos passageiros do Titanic armazenados em `../arquivos-csv/titanic.csv`.

> **Analogia de Engenharia:** Em ensaios laboratoriais, sensores frequentemente registram códigos especiais de erro (como `'?'` ou `-999`) quando falham. Antes de calcular a média ou o desvio padrão em um vetor `NumPy`, precisamos filtrar essas leituras inválidas (`NaN` — *Not a Number*, dado numérico ausente) para não contaminar toda a estatística!

## Passo 1 — Inspecionando os Dados Brutos
Execute a célula abaixo para carregar o arquivo `.csv` e observar como a coluna `age` chega originalmente e quantos registros válidos existem.

````python
# Importa o NumPy (cálculo vetorial) e o Pandas (leitura de tabelas CSV)
import numpy as np
import pandas as pd

# Lê o CSV subindo um nível de pasta ('..') e converte marcadores '?' em NaN
df = pd.read_csv("../arquivos-csv/titanic.csv", na_values=["?"])

# Converte a coluna 'age' para float e extrai como vetor 1D do NumPy (.to_numpy())
idades_brutas = pd.to_numeric(df["age"], errors="coerce").to_numpy()

# Conta total de linhas vs. quantas estão com NaN (np.isnan devolve True onde falta dado)
print(f"Total de passageiros registrados: {len(idades_brutas)}")
print(f"Idades ausentes (NaN): {np.sum(np.isnan(idades_brutas))}")
````

## Passo 2 — O Seu Desafio Prático

A partir do vetor `idades_brutas` carregado acima:
1. Crie um vetor limpo `idades` removendo os valores `NaN` com máscara booleana (`~np.isnan(idades_brutas)` — o operador `~` significa **NÃO lógico**, invertendo `True` e `False`).
2. Calcule com funções do `NumPy`: a **média** (`np.mean`), a **mediana** (`np.median`) e o **desvio padrão** (`np.std`) das idades.
3. Descubra a idade da pessoa **mais jovem** (`np.min`) e da **mais velha** (`np.max`) a bordo.
4. Conte quantos passageiros tinham **mais de 50 anos** e quantos eram **crianças com menos de 12 anos** usando soma de máscaras booleanas (`np.sum(idades > 50)`).

<details>
<summary><b>Dica rápida sobre máscaras booleanas</b></summary>

Quando você escreve `idades > 50`, o `NumPy` cria um vetor cheio de `True` e `False`. Em Python, `True` vale `1` e `False` vale `0`; por isso, `np.sum(idades > 50)` conta automaticamente quantos elementos satisfazem a condição sem precisar de laço `for`!

</details>

````python
# Escreva sua solução abaixo (comente cada passo para fixar o raciocínio!):
````

```python
import numpy as np
import pandas as pd

# 1. Lê o CSV subindo um nível de pasta ('..') e tratando '?' como NaN
df = pd.read_csv("../arquivos-csv/titanic.csv", na_values=["?"])

# 2. Garante tipo numérico e extrai a coluna como um vetor NumPy 1D
idades_brutas = pd.to_numeric(df["age"], errors="coerce").to_numpy()

# 3. Remove valores ausentes (NaN): ~np.isnan() mantém apenas números válidos!
idades = idades_brutas[~np.isnan(idades_brutas)]

# 4. Cálculos estatísticos vetorizados no NumPy
media = np.mean(idades)
mediana = np.median(idades)
desvio = np.std(idades)
minima = np.min(idades)
maxima = np.max(idades)

# 5. Contagem com máscaras booleanas (True = 1, False = 0 em np.sum)
acima_50 = int(np.sum(idades > 50.0))
menores_12 = int(np.sum(idades < 12.0))

print(f"Passageiros com idade conhecida: {len(idades)}")
print(f"Média: {media:.2f} anos | Mediana: {mediana:.2f} anos | Desvio Padrão: {desvio:.2f} anos")
print(f"Mais jovem: {minima:.2f} anos | Mais velho(a): {maxima:.1f} anos")
print(f"Maiores de 50 anos: {acima_50} | Crianças (< 12 anos): {menores_12}")
```

### O que observar nos resultados?
- Repare que a idade mínima é um número fracionário (`0.17` anos $\approx 2$ meses de vida — um bebê a bordo!).
- A **mediana** (`28.00` anos) é ligeiramente menor que a **média** (`29.88` anos), indicando uma leve cauda à direita causada por passageiros mais idosos (até `80` anos).
