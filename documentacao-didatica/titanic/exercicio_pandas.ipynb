## Taxa de sobrevivência por classe

O ` Titanic ` , um dos navios mais famosos da história, infelizmente naufragou em 1912, deixando muitas histórias de coragem e tragédia. Entre os dados coletados desse desastre, estão informações sobre os passageiros, como a classe do bilhete, idade, gênero e se sobreviveram ou não. Esses dados nos permitem investigar aspectos importantes, como se a classe socioeconômica influenciou as chances de sobrevivência.

Neste exercício, você utilizará um arquivo chamado ` titanic_survival.csv ` , que contém registros dos passageiros com, entre outras colunas, a classe ( ` pclass ` ) e um indicador de sobrevivência ( ` survived ` ), onde o valor ` 1 ` representa um passageiro que sobreviveu e ` 0 ` um que não sobreviveu.

Sua tarefa é criar um novo ` DataFrame ` que mostre a taxa de sobrevivência (em porcentagem) para cada uma das três classes de passageiro ( ` 1 ` , ` 2 ` e ` 3 ` ). Esse tipo de análise é útil para entender melhor o impacto das condições socioeconômicas no evento e pode revelar, por exemplo, se os passageiros da primeira classe tiveram maiores chances de sobrevivência em comparação com os da terceira classe.

    <summary> Dicas </summary>

    * Utilize ` pd.read_csv() ` para ler o arquivo ` titanic_survival.csv ` .
    * Use ` groupby() ` para agrupar os dados por ` pclass ` .
    * Calcule a média da coluna ` survived ` para cada grupo e multiplique por ` 100 ` para converter a fração em porcentagem.
    * Crie um novo ` DataFrame ` com as colunas ` pclass ` e ` survival_rate ` .
    * Utilize ` to_csv() ` para salvar o ` DataFrame ` final em ` survival_rate_by_class.csv ` , passando ` index=False ` para não incluir o índice.

```python
import pandas as pd

# 1. Lendo o arquivo CSV
df = pd.read_csv('titanic_survival.csv')

# 2. Calculando a taxa de sobrevivência por classe
survival_df = df.groupby('pclass')['survived'].mean().reset_index()
survival_df = survival_df.rename(columns={'survived': 'survival_rate'})
survival_df['survival_rate'] = survival_df['survival_rate'] * 100

# 3. Exibindo o resultado para conferência
print(survival_df)

# 4. Salvando o resultado em um novo arquivo CSV
survival_df.to_csv('survival_rate_by_class.csv', index=False)
```
