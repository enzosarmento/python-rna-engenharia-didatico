```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Carregando dados já convertendo '?' em NaN automaticamente
df = pd.read_csv("../arquivos-csv/titanic.csv", na_values=["?"])

# 2. Gráfico 1: Histograma de Idade separado por Sobrevivência
plt.figure(figsize=(9, 4.5))
sns.histplot(data=df, x="age", hue="survived", kde=True, element="step", palette="Set1")
plt.title("Distribuição de Idades por Status de Sobrevivência")
plt.xlabel("Idade (anos)")
plt.ylabel("Contagem de Passageiros")
plt.tight_layout()
plt.show()

# 3. Gráfico 2: Contagem por Gênero e Sobrevivência
plt.figure(figsize=(7, 4.5))
sns.countplot(data=df, x="sex", hue="survived", palette="pastel")
plt.title("Sobrevivência por Gênero no Titanic")
plt.xlabel("Gênero (female = mulher, male = homem)")
plt.ylabel("Quantidade de Passageiros")
plt.tight_layout()
plt.show()

# 4. Gráfico 3: Boxplot de Tarifas por Porto de Embarque (C=Cherbourg, Q=Queenstown, S=Southampton)
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df, x="embarked", y="fare", hue="embarked", palette="Set2", legend=False)
plt.ylim(0, 200)  # Foca na faixa até £ 200 para destacar as caixas interquartis
plt.title("Distribuição de Tarifas por Porto de Embarque")
plt.xlabel("Porto de Embarque")
plt.ylabel("Tarifa (£)")
plt.tight_layout()
plt.show()
```

### Leitura Visual dos Gráficos
1. **No Histograma:** Observe o pico acentuado de sobrevivência (`survived = 1`) na faixa de `0` a `8` anos — crianças pequenas tiveram prioridade clara nos botes salva-vidas, um comportamento **não linear** em relação à idade adulta!
2. **No Boxplot:** Passageiros que embarcaram em **Cherbourg (`C`)** pagaram tarifas medianas substancialmente maiores (maior concentração de 1ª classe) do que em **Queenstown (`Q`)** e **Southampton (`S`)**.
