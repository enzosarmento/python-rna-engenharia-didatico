# Exercício 02 — Taxa de Sobrevivência por Classe e Gênero com `Pandas`

Neste exercício vamos investigar quantitativamente como a classe socioeconômica (`pclass`: `1`, `2` ou `3`) e o gênero (`sex`) influenciaram as chances de sobrevivência (`survived`: `1` = sobreviveu, `0` = não sobreviveu) usando `../arquivos-csv/titanic_survival.csv`.

> **Analogia de Engenharia:** Agrupar dados com `groupby()` equivale a estratificar a produção de uma fábrica por turno ou linha de montagem para identificar onde está concentrada a taxa de falhas.

## Passo 1 — Explorando as Primeiras Linhas da Planilha
Execute o bloco abaixo para inspecionar as colunas disponíveis e os tipos de dados de `titanic_survival.csv`.

````python
# Importa a biblioteca Pandas para manipulação tabular (DataFrames)
import pandas as pd

# Carrega a tabela tratando eventuais '?' como dado ausente (NaN)
df = pd.read_csv("../arquivos-csv/titanic_survival.csv", na_values=["?"])

# Exibe as 5 primeiras linhas e confere as colunas relevantes
print("Dimensões da tabela (linhas, colunas):", df.shape)
display(df[["pclass", "survived", "sex", "age"]].head())
````

## Passo 2 — O Seu Desafio Prático

1. Usando `df.groupby('pclass')['survived'].mean()`, construa um novo `DataFrame` contendo as colunas `pclass` e `survival_rate` (taxa de sobrevivência **em porcentagem**, de `0` a `100`).
2. Salve esse resultado no arquivo `'survival_rate_by_class.csv'` usando `index=False` (para não gravar uma coluna extra de numeração lateral).
3. **Desafio Extra:** Faça um agrupamento duplo `df.groupby(['pclass', 'sex'])['survived'].mean() * 100` para medir o impacto combinado da classe e do gênero!

<details>
<summary><b>Dicas para começar</b></summary>

- Como `survived` contém apenas zeros (`0`) e uns (`1`), a média aritmética `.mean()` já calcula exatamente a proporção de sobreviventes! Basta multiplicar por `100.0` para obter a porcentagem.
- Use `.reset_index()` após `.mean()` para transformar o índice agrupado novamente em coluna comum e `.rename(columns={'survived': 'survival_rate'})` para renomear.

```python
import pandas as pd

# 1. Carrega os dados pelo caminho relativo correto e remove linhas sem 'pclass'/'survived'
df = pd.read_csv("../arquivos-csv/titanic_survival.csv", na_values=["?"])
df = df.dropna(subset=["pclass", "survived"])

# 2. Agrupa por classe, calcula a média de sobrevivência e converte em porcentagem (%)
survival_df = df.groupby("pclass")["survived"].mean().reset_index()
survival_df = survival_df.rename(columns={"survived": "survival_rate"})
survival_df["survival_rate"] = survival_df["survival_rate"] * 100.0

print("--- Taxa de Sobrevivência por Classe (%) ---")
print(survival_df.round(2))

# 3. Exporta o relatório limpo em CSV
survival_df.to_csv("survival_rate_by_class.csv", index=False)

# 4. Desafio Extra: Agrupamento cruzado (Classe + Gênero)
cruzado_df = (df.groupby(["pclass", "sex"])["survived"].mean() * 100.0).reset_index()
cruzado_df = cruzado_df.rename(columns={"survived": "survival_rate_pct"})
print("\n--- Taxa Combinada por Classe e Gênero (%) ---")
print(cruzado_df.round(2))
```

### Interpretação para Modelagem Neural
Observe o contraste dramático revelado pelo agrupamento duplo: mulheres da 1ª classe tiveram **mais de 96%** de sobrevivência, enquanto homens da 3ª classe ficaram perto de **15%**. Isso explica por que `sex` e `pclass` terão os maiores pesos sinápticos nos nossos modelos de Redes Neurais nos próximos notebooks!
