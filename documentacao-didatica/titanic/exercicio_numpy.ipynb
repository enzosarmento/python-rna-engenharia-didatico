## Exercício: Análise Estatística com NumPy

Agora que você já sabe como o ` NumPy ` facilita cálculos matemáticos em massa, vamos aplicar esse conhecimento no dataset do ` Titanic `. 

Embora o ` pandas ` seja ótimo para organizar os dados, o ` NumPy ` é quem faz o "trabalho pesado" de cálculo. Neste exercício, vamos focar na coluna de **idades** ( ` age ` ) dos passageiros para extrair algumas estatísticas fundamentais para qualquer análise de engenharia.

### O Desafio

Você deve carregar o arquivo ` titanic.csv `, extrair a coluna de idades e, utilizando **exclusivamente funções do NumPy**, responder:
1. Qual a média de idade dos passageiros?
2. Qual o desvio padrão das idades?
3. Qual a idade da pessoa mais velha a bordo?
4. Quantos passageiros tinham mais de 50 anos?

<details>
    <summary> Dicas </summary>

    * Use ` pd.read_csv() ` para carregar os dados, mas converta a coluna para NumPy com ` .values `.
    * Lembre-se que o dataset tem valores faltantes ( ` NaN ` ). O NumPy tem funções específicas para ignorar ` NaN ` em cálculos, como ` np.nanmean() `, ou você pode remover os ` NaN ` usando máscaras booleanas: ` idades = idades[~np.isnan(idades)] `.
    * Para contar quantos passageiros atendem a uma condição, você pode usar uma máscara booleana e a função ` np.sum() ` ou ` len() `.

```python
import pandas as pd
import numpy as np

# 1. Carregando os dados e extraindo a coluna como um array NumPy
df = pd.read_csv('titanic.csv')
df['age'] = pd.to_numeric(df['age'], errors='coerce') # Garante que são números
idades = df['age'].values

# 2. Limpando valores NaN (Not a Number) que impediriam os cálculos
idades = idades[~np.isnan(idades)]

# 3. Cálculos Estatísticos
media = np.mean(idades)
desvio = np.std(idades)
maxima = np.max(idades)

# 4. Filtragem e Contagem (Passageiros > 50 anos)
mais_de_50 = idades[idades > 50]
total_mais_de_50 = len(mais_de_50)

print(f"Média de Idade: {media:.2f} anos")
print(f"Desvio Padrão: {desvio:.2f}")
print(f"Idade Máxima: {maxima} anos")
print(f"Total de passageiros com mais de 50 anos: {total_mais_de_50}")
```
