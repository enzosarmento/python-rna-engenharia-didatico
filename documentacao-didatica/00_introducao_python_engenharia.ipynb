## 1. Variáveis, Comentários e Unidades Físicas

Na programação, uma **variável** é uma etiqueta colada em uma gaveta da memória onde guardamos um valor numérico ou textual.
- Tudo o que vem depois do símbolo `#` (cerquilha) é um **comentário**: o Python ignora esse texto, pois ele serve apenas para explicar o raciocínio humano (como anotações na margem do caderno).
- **Atenção crucial:** Em Python, o sinal `=` significa **atribuição** ("guarde o resultado da direita dentro da variável da esquerda"), e usamos ponto `.` (padrão internacional) como separador decimal, nunca vírgula!

````python
# Definindo grandezas de um ensaio de tração mecânica
forca_newton = 15000.0      # Força axial aplicada no corpo de prova (em N)
area_mm2 = 75.0             # Área da seção transversal (em mm²)
material = "Aço ABNT 1020"  # Texto (chamado de 'string' ou 'str', sempre entre aspas)

# Calculando a tensão normal: sigma = F / A (1 N/mm² = 1 MPa)
tensao_mpa = forca_newton / area_mm2  # Divide a força pela área e guarda em 'tensao_mpa'

# A função print() exibe informações na tela; o 'f' antes das aspas (f-string)
# permite inserir variáveis dentro de chaves {} formatando casas decimais (:.2f = 2 casas)
print(f"Material ensaiado: {material}")
print(f"Tensão normal calculada: {tensao_mpa:.2f} MPa")
````

**O que aconteceu acima?**
O Python avaliou a divisão `15000.0 / 75.0`, encontrou `200.0`, guardou em `tensao_mpa` e exibiu o relatório formatado. Repare nos operadores aritméticos fundamentais do Python:
- Soma (`+`), Subtração (`-`), Multiplicação (`*`), Divisão (`/`) e **Potenciação (`**`, ex.: `r ** 2` para $r^2$ — cuidado: `^` não é potência em Python!)**.

## 2. Coleções de Dados: Listas e Dicionários

Na engenharia, um sensor raramente faz uma única leitura; ele registra dezenas ou milhares de pontos ao longo do tempo.
1. **Lista (`list`):** Sequência ordenada de valores entre colchetes `[...]`. **Regra de ouro:** a contagem de posições (**índices**) em Python começa sempre no **`0`**!
2. **Dicionário (`dict`):** Ficha técnica entre chaves `{chave: valor}`, onde consultamos dados pelo nome da propriedade.

````python
# Lista com 5 leituras consecutivas de um termopar (em °C)
leituras_temp = [82.5, 84.0, 85.2, 89.7, 91.0]

# Acessando elementos pelo índice (lembre-se: o primeiro é o índice 0!)
primeira_leitura = leituras_temp[0]   # Pega o 1º valor (82.5)
ultima_leitura = leituras_temp[-1]    # Índice -1 é um atalho prático para o último valor (91.0)
quantidade = len(leituras_temp)       # len() conta quantos itens existem na lista

print(f"Total de leituras: {quantidade} | Inicial: {primeira_leitura} °C | Final: {ultima_leitura} °C")

# Dicionário representando a ficha técnica de uma bomba centrífuga
ficha_bomba = {
    "tag": "B-101A",          # Identificador da bomba na planta
    "vazao_m3h": 45.0,        # Vazão nominal em m³/h
    "rendimento": 0.78        # Eficiência hidráulica (78%)
}

# Consultamos o dicionário passando o nome da chave entre colchetes
print(f"Equipamento {ficha_bomba['tag']} opera com rendimento de {ficha_bomba['rendimento'] * 100:.1f}%")
````

Observe como listas organizam **séries de medições** (que logo transformaremos em vetores `NumPy`) e dicionários organizam **registros rotulados** (base das tabelas `Pandas`).

## 3. Tomada de Decisão (`if` / `elif` / `else`) e Indentação

Um sistema de controle industrial precisa comparar medidas com limites de segurança. Fazemos isso com `if` (*se*), `elif` (*senão, se*) e `else` (*caso contrário*).

> **Atenção à Indentação:** Em Python, os comandos subordinados a um `if` ficam **recuados 4 espaços para a direita** (basta apertar a tecla `Tab`). Esse alinhamento visual diz ao Python onde o bloco começa e termina!

````python
# Monitoramento de segurança da pressão de um vaso de pressão (em bar)
pressao_atual = 11.4          # Valor lido pelo manômetro digital
limite_operacional = 10.0     # Teto normal de operação (bar)
limite_critico = 12.0         # Pressão de abertura da válvula de alívio (PSV)

# Comparações usam >= (maior ou igual), <= (menor ou igual), == (igual) e != (diferente)
if pressao_atual >= limite_critico:
    # Só executa este bloco recuado se a pressão atingir ou passar de 12.0 bar
    print("ALARME CRÍTICO: Acionar válvula de alívio imediatamente!")
elif pressao_atual > limite_operacional:
    # Só executa se a primeira condição foi falsa, mas passou de 10.0 bar
    print("ATENÇÃO: Pressão acima da faixa nominal. Reduzir aquecimento.")
else:
    # Executa se todas as condições anteriores forem falsas
    print("OPERAÇÃO NORMAL: Vaso operando dentro dos limites de projeto.")
````

Como `pressao_atual` (`11.4`) é menor que `12.0`, mas maior que `10.0`, o programa entrou exatamente no ramo intermediário (`elif`). Nas Redes Neurais, o neurônio **Perceptron** toma decisões binárias usando exatamente essa lógica de limiar!

## 4. Repetição Automática com Laços (`for`)

Se precisarmos converter 5 temperaturas de Celsius para Kelvin ($T_K = T_C + 273{,}15$), não repetimos 5 linhas de código: usamos um laço `for` (*para cada elemento da lista, execute...*). No treinamento de Redes Neurais, cada ciclo completo de repetição sobre os dados será chamado de **Época (*Epoch*)**.

````python
# Lista de temperaturas medidas em um trocador de calor (°C)
temperaturas_celsius = [25.0, 50.0, 75.5, 100.0]

# Lista vazia onde acumularemos os valores convertidos para Kelvin
temperaturas_kelvin = []

# Para cada valor 'tc' dentro da lista 'temperaturas_celsius':
for tc in temperaturas_celsius:
    tk = tc + 273.15                  # Converte a leitura atual para Kelvin
    temperaturas_kelvin.append(tk)    # .append() adiciona o resultado ao final da nova lista
    print(f"{tc:6.2f} °C  -->  {tk:6.2f} K")

print("Lista final em Kelvin:", temperaturas_kelvin)
````

O laço `for` percorreu automaticamente os 4 itens da lista e construiu `temperaturas_kelvin` sem retrabalho manual.

## 5. Funções (`def`), Classes (`class`) e Bibliotecas (`import`)

Para fechar sua base de Python antes de entrarmos em Ciência de Dados, precisamos conhecer três estruturas que aparecerão em todos os notebooks:
1. **`def` (Função):** Empacota uma fórmula matemática reutilizável (ex.: potência elétrica $P = V \cdot I$).
2. **`class` (Classe):** É a "planta mecânica" que agrupa **parâmetros internos** (chamados de atributos, guardados em `self`) e **operações** (métodos). É assim que construímos uma Rede Neural do zero: `self.pesos` guarda a memória da rede e `treinar()` ajusta esses pesos!
3. **`import` (Importação):** Carrega módulos matemáticos prontos, como o módulo nativo `math` (seno, cosseno, exponencial $e^x$, $\pi$).

````python
import math  # Importa a biblioteca matemática padrão do Python

# 1. Criando uma função reutilizável para calcular a Lei de Ohm (V = R * I)
def calcular_tensao(resistencia_ohm, corrente_ampere):
    tensao = resistencia_ohm * corrente_ampere  # Aplica V = R * I
    return tensao                               # Devolve o valor calculado para quem chamou

v_circuito = calcular_tensao(resistencia_ohm=120.0, corrente_ampere=0.5)
print(f"Tensão no resistor: {v_circuito:.1f} V")

# 2. Criando uma classe simples que modela um Sensor Calibrável: y = ganho * x + offset
class SensorCalibrado:
    def __init__(self, ganho, offset):
        # '__init__' é o construtor: roda quando criamos o sensor e guarda seus parâmetros em 'self'
        self.ganho = ganho    # Equivalente ao 'peso' (w) de um neurônio artificial!
        self.offset = offset  # Equivalente ao 'viés / bias' de um neurônio artificial!

    def medir(self, sinal_bruto_mv):
        # Aplica a equação de reta de calibração usando os parâmetros guardados em 'self'
        return self.ganho * sinal_bruto_mv + self.offset

# Construindo um objeto 'termopar' a partir da planta 'SensorCalibrado'
termopar = SensorCalibrado(ganho=0.45, offset=-2.0)
print(f"Temperatura calibrada para 200 mV: {termopar.medir(200.0):.2f} °C")
print(f"Valor de pi vindo da biblioteca math: {math.pi:.5f}")
````

**Conexão direta com Redes Neurais:** Repare na classe `SensorCalibrado` acima! Um neurônio artificial nada mais é do que um objeto igualzinho a esse, onde `self.ganho` são os **pesos sinápticos** e `self.offset` é o **viés (*bias*)**, seguidos de uma curva de ativação (como `1 / (1 + math.exp(-u))`). A única diferença é que a Rede Neural **descobre sozinha** os melhores valores de ganho e offset a partir dos dados experimentais!
