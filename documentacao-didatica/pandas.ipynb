# Primeiros passos com o ` pandas ` : Análise do DataSet ` Titanic `

Vamos começar a explorar o dataset ` Titanic ` usando o ` pandas ` . Esse dataset contém informações sobre os passageiros do famoso navio ` Titanic ` , que naufragou em 1912. O nosso objetivo é entender como os dados estão organizados e o que podemos aprender sobre os passageiros.

### Importando o ` pandas ` e lendo o arquivo ` CSV `

Aqui, estamos usando a palavra-chave ` import ` para importar a biblioteca ` pandas ` e ` as pd ` para dar a ela o apelido de ` pd ` . Esse é um padrão comum na comunidade ` Python ` , e usaremos ` pd ` para nos referir ao ` pandas ` daqui em diante.

Em seguida, ` pd.read_csv() ` é a função do ` pandas ` que lê arquivos ` CSV ` (Comma Separated Values, ou Valores Separados por Vírgula). Os arquivos ` CSV ` são uma forma comum de armazenar dados tabulares em formato de texto. O resultado da leitura é armazenado em uma variável chamada ` df ` , que é um objeto do tipo ` DataFrame ` .

```python
import pandas as pd

df = pd.read_csv('titanic.csv')
```

* **` DataFrame `** : Um ` DataFrame ` é a estrutura de dados principal do ` pandas ` , que representa uma tabela de dados, onde as linhas representam as observações (no nosso caso, passageiros do ` Titanic ` ) e as colunas representam as features (características dos passageiros, como idade, sexo, classe, etc.). Você pode pensar em um ` DataFrame ` como uma matriz, conceito que você já está familiarizado, mas com funcionalidades adicionais que facilitam a manipulação e análise dos dados.

### Visualizando os dados:
O método ` head() ` é uma forma rápida de visualizar as primeiras linhas de um ` DataFrame ` . Por padrão, ele mostra as 5 primeiras linhas, mas você pode especificar um número diferente, como iremos fazer abaixo, passando o número desejado entre parênteses.

```python
print(df.head(10))
```

O método ` tail() ` é parecido com o ` head() ` , mas diferentemente do ` head() ` mostra as últimas linhas do ` DataFrame ` .

```python
print(df.tail(7))
```

### Obter informações úteis sobre o ` DataFrame ` :

O método ` info() ` fornece informações úteis sobre o ` DataFrame ` , como:
* Número total de linhas e colunas.
* Os nomes das colunas.
* O tipo de dado de cada coluna.
* A quantidade de memória usada pelo ` DataFrame ` .
* A quantidade de valores não nulos em cada coluna.

Podemos ver sua execução abaixo:

```python
print(df.info())
```

É importante para identificarmos a estrutura de nossos dados e identificar se existe valores faltantes. Segundo as informações extraídas do método ` info() ` , temos 1309 dados de passageiros e todas as colunas têm 1309 valores, ou seja, nenhum valor nulo.

### Estatísticas descritivas:

O método ` describe() ` calcula várias estatísticas para cada coluna numérica do ` DataFrame ` , como:
* **` count `** : A quantidade de valores não nulos.
* **` mean `** : A média dos valores.
* **` std `** : O desvio padrão, que mede a dispersão dos dados em relação à média.
* **` min `** : O menor valor.
* **` 25% `** : O primeiro quartil (25% dos dados estão abaixo desse valor).
* **` 50% `** : A mediana (50% dos dados estão abaixo desse valor).
* **` 75% `** : O terceiro quartil (75% dos dados estão abaixo desse valor).
* **` max `** : O maior valor.

Essas estatísticas nos dão uma ideia geral da distribuição dos valores em cada coluna numérica e podem nos ajudar a identificar outliers (valores extremos) ou padrões interessantes nos dados.

Podemos ver a execução do método ` describe() ` abaixo:

```python
print(df.describe())
```

## Manipulando Dados e Selecionando Features

Agora que já temos uma noção básica do nosso dataset, vamos começar a manipulá-lo para prepará-lo para a construção do nosso modelo de ` IA ` . Essa etapa de preparação é crucial em qualquer projeto de ` Machine Learning ` , pois a qualidade dos dados impacta diretamente na performance do modelo.

### Lidando com valores faltantes

Valores faltantes são um problema comum em datasets reais. Eles podem ocorrer por diversos motivos, como erros na coleta de dados ou informações incompletas. É importante tratá-los adequadamente, pois muitos algoritmos de ` Machine Learning ` não lidam bem com dados faltantes.

### Identificando e tratando valores faltantes

Quando executamos a função ` df.info() ` , não identificamos nenhum valor nulo diretamente. No entanto, ao abrir o arquivo ` CSV ` , é possível perceber que vários valores estão preenchidos com ` '?' ` . Esse tipo de situação é comum quando trabalhamos com conjuntos de dados reais e demonstra a importância de analisar os dados de várias formas para evitar problemas futuros.

Para padronizar a representação de valores faltantes, vamos substituir esses valores ` '?' ` por ` NaN ` (Not a Number), que é a representação padrão do ` pandas ` para valores ausentes.

Para isso, iremos utilizar o método ` replace() ` do ` pandas ` para substituir todas as ocorrências de ` "?" ` por ` pd.NA ` . O que é esse ` pd.NA ` ? ` pd.NA ` é a representação do ` pandas ` para "Not a Number" ( ` NaN ` ). Ele foi introduzido como uma forma mais consistente de lidar com dados faltantes, especialmente em colunas com tipos de dados mistos.

Podemos ver a execução do método ` replace() ` abaixo:

```python
df = df.replace('?', pd.NA)
```

### Preenchendo valores faltantes

Agora que identificamos e padronizamos os valores faltantes, precisamos decidir como tratá-los. Antes de prosseguir, vamos ver quais colunas têm valores ` NaN ` . Para isso, usaremos o método ` df.isna() ` seguido de ` any() ` :

```python
columns_with_nan = df.isna().any() # Retorna True para colunas com valores ausentes

print(columns_with_nan)
```

Analisando a execução acima, podemos ver que todos os campos com o valor ` pd.NA ` foram substituídos por ` True ` , enquanto os outros valores foram substituídos por ` False ` . A função ` any() ` simplifica o resultado para identificar em cada coluna se existe um valor ` pd.NA ` nessa coluna. Com isso foi possível indentificar que as colunas ` "age" ` , ` "fare" ` , ` "cabin" ` , ` "embarked" ` , ` "boat" ` , ` "body" ` e ` "home.dest" ` têm valores faltantes, e precisamos decidir o que fazer com essa informação.

Para tratar esses valores existem várias estratégias, e a escolha depende do contexto e da quantidade de dados faltantes.

* Para variáveis numéricas: Uma estratégia comum é preencher os valores faltantes com a média da coluna. Vamos fazer isso para as colunas ` "age" ` e ` "fare" ` .
* Para variáveis categóricas: Uma abordagem comum é preencher os valores faltantes com o valor mais frequente (moda). Vamos fazer isso para a coluna ` "embarked" ` . Poderíamos fazer para as outras colunas? Sim, essa decisão tem que ser feita baseada no seu entendimento dos dados. Por exemplo, caso usássemos a moda para preencher os dados da coluna ` cabin ` , poderíamos estar colocando dezenas de pessoas em uma só cabine. Isso afetará nosso modelo no final? Não sabemos, mas podemos voltar nesse ponto posteriormente e revisar quais colunas queremos tratar.

```python
numerical_features = ["age", "fare"]
for column in numerical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mean())

categorical_features = ["embarked"]
for column in categorical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mode()[0])
```

Este código preenche valores ausentes ( ` NaN ` ) no ` DataFrame ` ` df ` de forma específica para colunas numéricas e categóricas. Vamos detalhar passo a passo:

* **` numerical_features `** : Uma lista que contém os nomes das colunas numéricas que queremos processar.
* **` for column in numerical_features `** : Percorre cada nome de coluna listado em ` numerical_features ` .
* **` if column in df.columns `** : Verifica se a coluna realmente existe no ` DataFrame ` ` df ` . Isso evita erros caso a coluna especificada não esteja presente.
* **` df[column].mean() `** : Calcula a média dos valores existentes na coluna (ignora valores ausentes automaticamente).
    * Exemplo: Se a coluna ` "age" ` tem os valores ` [25, 30, NaN, 35] ` , a média será ` 30 ` .

* **` df[column].fillna(df[column].mean()) `** :
    * Substitui todos os valores ausentes na coluna pela média calculada.
    * Exemplo: Após o preenchimento, a coluna ` "age" ` ficaria ` [25, 30, 30, 35] ` .

* Por fim, ` df[column] = df[column].fillna(df[column].mean()) ` faz com que a coluna com nome ` column ` seja substituída por uma nova coluna. No caso, uma coluna com todos os valores ` NaN ` substituídos pela média dos outros valores na coluna.

O mesmo é feito para as colunas categóricas, porém, agora usando a moda:

* **` df[column].mode() `** : Calcula a moda da coluna, ou seja, o valor mais frequente.
    * Exemplo: Se a coluna ` "embarked" ` tem os valores ` ["S", "C", "S", NaN] ` , a moda será ` "S" ` .

* **` df[column].mode()[0] `** :
    * Como ` mode() ` pode retornar múltiplos valores (em caso de empate), usamos ` [0] ` para pegar o primeiro valor.

* **` df[column].fillna(df[column].mode()[0]) `** :
    * Substitui todos os valores ausentes na coluna pelo valor mais frequente.
    * Exemplo: Após o preenchimento, a coluna ` "embarked" ` ficaria ` ["S", "C", "S", "S"] ` .

### Erro nos dados

Ao tentar executar o código acima, irá ocorrer um erro.

Esse erro está dizendo que não é possível somar inteiros com strings. O único local no nosso código em que esse tipo de cálculo pode estar sendo feito é no cálculo da média. E isso quer dizer que alguns valores de alguma coluna que deveriam ser inteiros estão codificados como strings. Vamos tentar descobrir quais colunas contêm esse erro.

Para isso, vamos capturar a exceção e mostrar o nome da coluna em que esse problema acontece:

```python
numerical_features = ["age", "fare"]
for column in numerical_features:
    if column in df.columns:
        try:
            df[column] = df[column].fillna(df[column].mean())
        except TypeError as e:
            print(column, e)

categorical_features = ["embarked"]
for column in categorical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mode()[0])
```

Ao adicionar o bloco ` try...except ` acima, nós descobrimos que tanto a coluna ` age ` como ` fare ` contêm strings quando deveriam ser números.

Vamos forçar a conversão dos valores de string para inteiros usando o método ` pd.to_numeric() ` e com isso podemos remover o bloco ` try...except ` , o código ficando da seguinte maneira:

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["fare"] = pd.to_numeric(df["fare"], errors="coerce")

numerical_features = ["age", "fare"]
for column in numerical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mean())

categorical_features = ["embarked"]
for column in categorical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mode()[0])

print(df) # Imprime o estado atual do DataFrame, após as modificações
```

## Resumo

Após realizar as operações acima, nós conseguimos entender um pouco sobre os valores presentes no ` dataset ` , conseguimos identificar valores faltantes, além de corrigir valores com codificação errada. Nosso código final ficou o seguinte:

```python
import pandas as pd

# Carrega os dados do arquivo CSV do Titanic
df = pd.read_csv("titanic.csv")

# Mostra as 10 primeiras linhas do DataFrame
print(df.head(10)) 

# Mostra as 7 últimas linhas do DataFrame
print(df.tail(7))

# Exibe informações gerais sobre o DataFrame, como tipos de dados e valores nulos
print(df.info())

# Exibe estatísticas descritivas para as colunas numéricas do DataFrame
print(df.describe())

# Substitui valores "?" por valores nulos reconhecidos pelo pandas
df = df.replace("?", pd.NA)

# Identifica colunas com valores ausentes (NaN) no DataFrame
columns_with_nan = df.isna().any()

# Converte as colunas "age" e "fare" para numéricas, tratando erros como valores nulos
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["fare"] = pd.to_numeric(df["fare"], errors="coerce")

# Preenche valores ausentes em colunas numéricas com a média da coluna
numerical_features = ["age", "fare"]
for column in numerical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mean())

# Preenche valores ausentes em colunas categóricas com o valor mais frequente (moda)
categorical_features = ["embarked"]
for column in categorical_features:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mode()[0])

# Exibe o DataFrame atualizado com os valores ausentes tratados
print(df)
```
