# Classificando as espécies de ` Íris `

O conjunto de dados ` IRIS ` é amplamente utilizado em aprendizado de máquina e reúne informações sobre três espécies de flores: *setosa*, *versicolor* e *virginica*. Cada registro contém quatro características: comprimento e largura das sépalas, além do comprimento e largura das pétalas.

Neste exercício, o objetivo é desenvolver um modelo capaz de classificar corretamente a espécie de cada flor com base nesses atributos.

Você terá acesso a dois arquivos:

- ` iris_train.csv `: contém os dados completos, incluindo as quatro variáveis de entrada ( ` sepal_length `, ` sepal_width `, ` petal_length `, ` petal_width ` ) e a coluna *class*, que indica a espécie da flor. Esse arquivo deve ser utilizado para treinar o modelo.
- ` iris_exercise.csv `: possui apenas os atributos de entrada, sem a coluna *class*. O modelo treinado deverá ser utilizado para prever a espécie de cada amostra presente nesse arquivo.

## Desafio
1. **Treinamento do modelo**: utilize os dados do arquivo ` iris_train.csv ` para treinar um modelo de classificação capaz de identificar corretamente a espécie da íris.
2. **Geração de previsões**: aplique o modelo treinado sobre os dados do arquivo ` iris_exercise.csv ` para prever a classe de cada amostra.
3. **Criação do arquivo**: gere um arquivo chamado ` solution.csv ` contendo o resultado das previsões no seguinte formato:

```csv
id,class
1,Iris-setosa
2,Iris-versicolor
3,Iris-virginica
...
```

Após implementar sua solução, você poderá ver as diversas formas de resolver o problema abaixo:

````python
# Escreva seu código aqui
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler

# 1. Carregar e Limpar
train = pd.read_csv('../arquivos-csv/iris_train.csv')
test = pd.read_csv('../arquivos-csv/iris_exercise.csv')

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X_train, y_train = train[features], train['class']
X_test = test[features]

# 2. Normalizar e Treinar
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
clf = Perceptron(random_state=42)
clf.fit(X_train_scaled, y_train)

# 3. Salvar Resposta
y_pred = clf.predict(scaler.transform(X_test))
pd.DataFrame({'id': test['id'], 'class': y_pred}).to_csv('solution.csv', index=False)
print("CSV gerado: Perceptron (Scikit-Learn)")
````

#### 1.1 Perceptron Simples - Utilizando Scikit-Learn

````python
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler

# 1. Carregar e Limpar
train = pd.read_csv('../arquivos-csv/iris_train.csv')
test = pd.read_csv('../arquivos-csv/iris_exercise.csv')

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X_train, y_train = train[features], train['class']
X_test = test[features]

# 2. Normalizar e Treinar
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
clf = Perceptron(random_state=42)
clf.fit(X_train_scaled, y_train)

# 3. Salvar Resposta
y_pred = clf.predict(scaler.transform(X_test))
pd.DataFrame({'id': test['id'], 'class': y_pred}).to_csv('solution_iris_perceptron_sklearn.csv', index=False)
print("CSV gerado: Perceptron (Scikit-Learn)")
````

#### 1.2 Perceptron Simples - Utilizando Python Puro

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Carregar e Preparar
train = pd.read_csv('../arquivos-csv/iris_train.csv')
test = pd.read_csv('../arquivos-csv/iris_exercise.csv')

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X_train_raw = train[features].values
y_train_raw = train['class'].values
X_test_raw = test[features].values

# O Perceptron Puro precisa de classes numéricas e normalização ajuda muito
classes = list(np.unique(y_train_raw))
y_train = np.array([classes.index(c) for c in y_train_raw])

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_raw)
X_test = scaler.transform(X_test_raw)

# 2. Algoritmo Perceptron (Multiclasse simplificado)
weights = np.zeros((len(classes), X_train.shape[1] + 1))
lr, epochs = 0.01, 100
for _ in range(epochs):
    for i in range(len(X_train)):
        x = np.insert(X_train[i], 0, -1)
        for c_idx in range(len(classes)):
            target = 1 if y_train[i] == c_idx else 0
            y_hat = 1 if np.dot(weights[c_idx], x) >= 0 else 0
            weights[c_idx] += lr * (target - y_hat) * x

# 3. Previsão e CSV
preds = []
for x_row in X_test:
    scores = [np.dot(weights[c], np.insert(x_row, 0, -1)) for c in range(len(classes))]
    preds.append(classes[np.argmax(scores)])

pd.DataFrame({'id': test['id'], 'class': preds}).to_csv('solution_iris_perceptron_puro.csv', index=False)
print("CSV gerado: Perceptron (Puro)")
````

#### 2.1 MLP (Rede Neural) - Utilizando Scikit-Learn

#### 2.2 MLP (Rede Neural) - Utilizando Python Puro

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Carregar e Preparar
train = pd.read_csv('../arquivos-csv/iris_train.csv')
test = pd.read_csv('../arquivos-csv/iris_exercise.csv')

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X_train_raw = train[features].values
y_train_raw = train['class'].values
X_test_raw = test[features].values

classes = list(np.unique(y_train_raw))
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

class MLP_Iris:
    def __init__(self, d_in, d_hid, d_out):
        np.random.seed(42)
        self.w1 = np.random.randn(d_in, d_hid)
        self.w2 = np.random.randn(d_hid, d_out)
        self.lr = 0.05
    def train(self, X, y_onehot, ep=2000):
        for _ in range(ep):
            h = 1 / (1 + np.exp(-np.dot(X, self.w1)))
            o = 1 / (1 + np.exp(-np.dot(h, self.w2)))
            do = (y_onehot - o) * (o * (1 - o))
            dh = do.dot(self.w2.T) * (h * (1 - h))
            self.w2 += h.T.dot(do) * self.lr
            self.w1 += X.T.dot(dh) * self.lr
    def pred(self, X):
        h = 1 / (1 + np.exp(-np.dot(X, self.w1)))
        o = 1 / (1 + np.exp(-np.dot(h, self.w2)))
        return np.argmax(o, axis=1)

# One-Hot Encoding para as 3 classes
y_onehot = pd.get_dummies(y_train_raw).values

# Treinar e Salvar
net = MLP_Iris(X_train_scaled.shape[1], 12, 3)
net.train(X_train_scaled, y_onehot)
y_res_idx = net.pred(X_test_scaled)
y_res = [classes[i] for i in y_res_idx]

pd.DataFrame({'id': test['id'], 'class': y_res}).to_csv('solution_iris_mlp_puro.csv', index=False)
print("CSV gerado: MLP (Puro)")
````
